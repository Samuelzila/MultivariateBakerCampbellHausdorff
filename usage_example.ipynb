```python
from bch import MultivariateBakerCampbellHausdorff
from sage.algebras.lie_algebras.lie_algebra import LieAlgebra
```

```python
# A lie algebra over the symbolic ring SR, generated by H0, H1 and H2.
g = LieAlgebra(SR, "H", 3) 

bch = MultivariateBakerCampbellHausdorff(g, g.gens())

bch(2)
```

```python
from sage.matrix.matrix_space import MatrixSpace
```

```python
MS = MatrixSpace(SR, 2)
# Create a Lie algebra corresponding to an associative algebra.
matrix_lie_algebra = LieAlgebra(associative=MS.gens())

matrix_bch = MultivariateBakerCampbellHausdorff(
    matrix_lie_algebra,
    [
        MS([
            [1,2],
            [3,4]
        ]),
        MS([
            [5,6],
            [7,8]
        ])
    ]
                                               )

matrix_bch.nth(2)
```
